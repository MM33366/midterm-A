````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def init_db():
    conn = sqlite3.connect('events.db')
    try:
        cursor = conn.cursor()
        cursor.execute("DROP TABLE IF EXISTS registrations")
        cursor.execute("DROP TABLE IF EXISTS events")

        cursor.execute("""
            CREATE TABLE events (
                name TEXT PRIMARY KEY,
                date TEXT,
                capacity INTEGER
            )
        """)

        cursor.execute("""
            CREATE TABLE registrations (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT,
                event_name TEXT,
                party_size INTEGER,
                status TEXT
            )
        """)

        initial_events = [
            ("迎新晚會", "2026-11-20", 6),
            ("程式工作坊", "2026-11-25", 4),
            ("籃球友誼賽", "2026-11-28", 3)
        ]

        cursor.executemany("INSERT INTO events (name, date, capacity) VALUES (?, ?, ?)", initial_events)
        conn.commit()
        print("✅ 資料庫初始化完成，已寫入初始活動資料。")
    finally:
        conn.close()

init_db()
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def show_tables():
    conn = sqlite3.connect('events.db')
    try:
        print("【活動表 events】")
        df_events = pd.read_sql_query("SELECT * FROM events", conn)
        display(df_events)

        print("\n【報名紀錄 registrations】")
        df_regs = pd.read_sql_query("SELECT * FROM registrations", conn)
        if df_regs.empty:
            print("（目前無任何報名紀錄）")
        else:
            display(df_regs)
    finally:
        conn.close()

show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
ai_cache = {}

def parse_intent(text):
    if text in ai_cache:
        return ai_cache[text]

    system_prompt = """
    你是一個系學會活動報名小幫手。請從使用者的話中擷取「動作」、「活動」與「人數」。
    - 動作只能是：「報名」、「取消」、「查詢」、「統計」。如果都不是，請輸出「未知」。
      * 查詢：想知道「特定單一活動」還剩多少名額。
      * 統計：想知道「所有活動」或「各活動」的報名狀況。
    - 活動只能是：「迎新晚會」、「程式工作坊」、「籃球友誼賽」。如果都沒提到，請輸出「未知」。如果使用者問「各活動狀況」，活動請輸出「未知」。
    - 人數請提取數字（例如我和兩個朋友 = 3）。如果沒提到數量，預設為 1。
    請嚴格遵守輸出格式，只要輸出三個欄位，用逗號分隔，例如：報名,迎新晚會,3
    """

    retries = 3
    for attempt in range(retries):
        try:
            response = client.chat.completions.create(
                model=MODEL,
                messages=[
                    {"role": "system", "content": system_prompt},
                    {"role": "user", "content": text}
                ],
                temperature=0,
            )
            raw_reply = response.choices[0].message.content.strip()

            allowed_actions = ["報名", "取消", "查詢", "統計"]
            allowed_events = ["迎新晚會", "程式工作坊", "籃球友誼賽"]

            action = "未知"
            event = "未知"
            size = 1

            for a in allowed_actions:
                if a in raw_reply:
                    action = a
                    break

            for e in allowed_events:
                if e in raw_reply:
                    event = e
                    break

            match = re.search(r'\d+', raw_reply)
            if match:
                size = int(match.group())

            result = (action, event, size, raw_reply)
            ai_cache[text] = result
            time.sleep(2)
            return result

        except Exception as e:
            err_msg = str(e)
            if "429" in err_msg:
                print(f"⚠️ API 請求過於頻繁，等待 30 秒後重試 ({attempt+1}/{retries})...")
                time.sleep(30)
            else:
                print(f"❌ AI 呼叫發生錯誤: {e}")
                return ("未知", "未知", 1, err_msg)

    return ("未知", "未知", 1, "API Timeout")
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
test_cases = [
    ("幫我報名迎新晚會", "報名", "迎新晚會", 1),
    ("我和朋友兩個人要參加程式工作坊", "報名", "程式工作坊", 2),
    ("取消籃球友誼賽", "取消", "籃球友誼賽", 1),
    ("迎新晚會還剩多少位子", "查詢", "迎新晚會", 1),
    ("顯示各活動報名狀況", "統計", "未知", 1),
    ("我們四個人要報名程式工作坊", "報名", "程式工作坊", 4),
    ("幫我報名期末聚餐", "報名", "未知", 1),
    ("今天天氣真好", "未知", "未知", 1)
]

passed = 0
print("【AI 解讀測試】")
for text, exp_act, exp_evt, exp_size in test_cases:
    act, evt, size, raw = parse_intent(text)
    is_pass = (act == exp_act) and (evt == exp_evt) and (size == exp_size)
    if is_pass:
        passed += 1
    status = "✅" if is_pass else "❌"
    print(f"{status} 輸入: {text}")
    print(f"   預期: ({exp_act}, {exp_evt}, {exp_size}) | 系統解讀: ({act}, {evt}, {size}) | 原始: {raw}")

print(f"\n總結：通過 {passed} / {len(test_cases)} 句")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def get_available_events():
    return "目前有的活動：迎新晚會 (11-20)、程式工作坊 (11-25)、籃球友誼賽 (11-28)。"

def handle_register(name, event, size):
    if not name:
        return "請提供姓名以進行報名。"
    if size < 1 or size > 3:
        return f"報名失敗：每次報名人數必須介於 1 到 3 人之間 (您要求 {size} 人)。"

    conn = sqlite3.connect('events.db')
    try:
        cursor = conn.cursor()
        cursor.execute("SELECT capacity FROM events WHERE name = ?", (event,))
        row = cursor.fetchone()
        if not row:
            return f"找不到活動「{event}」。{get_available_events()}"
        capacity = row[0]

        cursor.execute("SELECT id FROM registrations WHERE name = ? AND event_name = ? AND status = '報名成功'", (name, event))
        if cursor.fetchone():
            return f"報名失敗：您已經報名過「{event}」了，不可重複報名。"

        cursor.execute("SELECT COUNT(*) FROM registrations WHERE name = ? AND status = '報名成功'", (name,))
        if cursor.fetchone()[0] >= 2:
            return "報名失敗：每位學生最多只能同時報名 2 個活動。"

        cursor.execute("SELECT COALESCE(SUM(party_size), 0) FROM registrations WHERE event_name = ? AND status = '報名成功'", (event,))
        enrolled = cursor.fetchone()[0]
        if enrolled + size > capacity:
            return f"報名失敗：「{event}」剩餘名額不足 ({capacity - enrolled} 位)，無法報名 {size} 人。"

        cursor.execute("INSERT INTO registrations (name, event_name, party_size, status) VALUES (?, ?, ?, '報名成功')", (name, event, size))
        conn.commit()
        return f"報名成功！已為 {name} 登記 {event} 共 {size} 人。"
    finally:
        conn.close()

def handle_cancel(name, event):
    if not name:
        return "請提供姓名以進行取消。"

    conn = sqlite3.connect('events.db')
    try:
        cursor = conn.cursor()
        cursor.execute("SELECT id FROM registrations WHERE name = ? AND event_name = ? AND status = '報名成功'", (name, event))
        row = cursor.fetchone()
        if not row:
            return f"找不到 {name} 報名「{event}」的成功紀錄。"

        cursor.execute("UPDATE registrations SET status = '已取消' WHERE id = ?", (row[0],))
        conn.commit()
        return f"取消成功！已取消 {name} 的 {event} 報名，名額已釋出。"
    finally:
        conn.close()

def handle_query(event):
    conn = sqlite3.connect('events.db')
    try:
        cursor = conn.cursor()
        cursor.execute("SELECT capacity FROM events WHERE name = ?", (event,))
        row = cursor.fetchone()
        if not row:
            return f"找不到活動「{event}」。{get_available_events()}"

        capacity = row[0]
        cursor.execute("SELECT COALESCE(SUM(party_size), 0) FROM registrations WHERE event_name = ? AND status = '報名成功'", (event,))
        enrolled = cursor.fetchone()[0]
        return f"「{event}」剩餘名額為：{capacity - enrolled} 人。"
    finally:
        conn.close()

def handle_stats():
    conn = sqlite3.connect('events.db')
    try:
        query = """
            SELECT e.name, e.date, e.capacity, COALESCE(SUM(r.party_size), 0) as enrolled,
                   (e.capacity - COALESCE(SUM(r.party_size), 0)) as remaining
            FROM events e
            LEFT JOIN registrations r ON e.name = r.event_name AND r.status = '報名成功'
            GROUP BY e.name, e.date, e.capacity
        """
        df = pd.read_sql_query(query, conn)

        res = "📊 【各活動報名狀況】\n"
        for _, row in df.iterrows():
            res += f"- {row['name']} ({row['date']}): 總名額 {row['capacity']} | 已報名 {row['enrolled']} | 剩餘 {row['remaining']}\n"
        return res.strip()
    finally:
        conn.close()
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def process_request(name, text):
    act, evt, size, raw = parse_intent(text)

    if act == "報名":
        if evt == "未知": return get_available_events()
        return handle_register(name, evt, size)

    elif act == "取消":
        if evt == "未知": return get_available_events()
        return handle_cancel(name, evt)

    elif act == "查詢":
        if evt == "未知": return get_available_events()
        return handle_query(evt)

    elif act == "統計":
        return handle_stats()

    else:
        return f"無法理解您的需求。{get_available_events()}"

init_db()

steps = [
    ("王小明", "幫我報名迎新晚會"),
    ("王小明", "我還要報名迎新晚會"),
    ("王小明", "我和朋友共3人報名程式工作坊"),
    ("王小明", "幫我報名籃球友誼賽"),
    ("李小華", "我要帶4個朋友報名迎新晚會"),
    ("王小明", "幫我取消迎新晚會"),
    ("李小華", "各活動報名狀況"),
    ("張大山", "程式工作坊還有名額嗎")
]

print("【流程測試】")
for i, (p_name, p_text) in enumerate(steps, 1):
    print(f"\n[{i}] {p_name} 說: {p_text}")
    res = process_request(p_name, p_text)
    print(f"系統回應: {res}")

print("\n【最終資料表狀態】")
show_tables()
````
